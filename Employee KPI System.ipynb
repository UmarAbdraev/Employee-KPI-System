```python
employees = [
    "Алексей Иванов",
    "Мария Петрова",
    "Данияр Сарсенов",
    "Елена Ковалева",
    "Артем Сидоров"
]

sprints = [
    [85.0, 92.0, 78.0, 95.0, 88.0, 90.0],
    [70.0, 65.0, 80.0],
    [95.0, 100.0, 98.0, 92.0],
    [],
    [60.0, 75.0, 82.0, 90.0, 68.0]
]

projects = [
    90.0,
    85.0,
    None,
    70.0,
    None
]

audits = [
    95.0,
    None,
    88.0,
    60.0,
    None
]
```

```python
def add_employee(name):
    employees.append(name)
    sprints.append([])
    projects.append(None)
    audits.append(None)

def remove_employee(name):
    for i, v in enumerate(employees):
        if v == name:
            employees.pop(i)
            sprints.pop(i)
            projects.pop(i)
            audits.pop(i)

def add_sprint_score(name, score):
    if name in employees and (score >= 0 and score <= 100):
        for i, v in enumerate(employees):
            if v == name:
                sprints[i].append(score)
    elif name not in employees:
        return f'{name} does not exists in {employees}'  
    else:
        return f'{score} is not in diapozone 0-100'          

def set_project_score(name, score):
    if name in employees and (score >= 0 and score <= 100):
        for i, v in enumerate(employees):
            if v == name:
                projects[i] = score
    elif name not in employees:
            return f'{name} does not exists in {employees}'  
    else:
        return f'{score} is not in diapozone 0-100'   

def set_audit_score(name, score):
    if name in employees and (score >= 0 and score <= 100):
        for i, v in enumerate(employees):
            if v == name:
                audits[i] = score
    elif name not in employees:
            return f'{name} does not exists in {employees}'  
    else:
        return f'{score} is not in diapozone 0-100'   

def get_sprint_score(name):
    if name in employees:
        idx = employees.index(name)
        score = sprints[idx]
        if not score:  # Если список спринтов пустой []
            return 0.0
        
        if len(score) >= 4:
            x = sorted(score)
            return (sum(x[-4:]) / 4) * 0.5
        else:
            return (sum(score) / len(score)) * 0.5
    return 0.0

def get_perfomance_tier(score):
    if score >= 90:
        return 'Грейд: A (Премия: 100%)'
    elif score >= 80 and score < 90:
        return 'Грейд: B (Премия: 50%)'
    elif score >= 70 and score < 80:
        return 'Грейд: C (Без премии)'
    else:
        return 'Грейд: F (План улучшений)'

def calculate_total_kpi(name):
    if name in employees:
        for i, v in enumerate(employees):
            if v == name:
                if projects[i] != None and audits[i] != None:
                    x = sorted(sprints[i])
                    avg = sum(x[-4:]) / 4
                    total = (avg) * 0.5 + projects[i] * 0.3 + audits[i] * 0.2
                    return f'Итог: {total} | {get_perfomance_tier(total)}'
                else:
                    return f'{name} НЕ АТТЕСТОВАН.'

def report_KPI(name):
    if name in employees:
        idx = employees.index(name)
        
        p_val = projects[idx] if projects[idx] is not None else 0.0
        a_val = audits[idx] if audits[idx] is not None else 0.0
        
        sprint_pts = get_sprint_score(name)
        project_pts = p_val * 0.3
        audit_pts = a_val * 0.2
        
        status_kpi = calculate_total_kpi(name)
        
        return f"[ОТЧЕТ KPI] Сотрудник: {name} | Спринты(Top4): {sprint_pts:.2f}/50 | Проект: {project_pts:.2f}/30 | Аудит: {audit_pts:.2f}/20 | {status_kpi}"
```

```python
print(report_KPI('Артем Сидоров'))
```
